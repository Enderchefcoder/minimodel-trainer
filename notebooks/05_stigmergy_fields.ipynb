# 05 — Visualize Stigmergy pheromone fields

Deposit + diffuse the 10-channel lattice on the start position and a tactical middlegame.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().resolve()
if not (ROOT / "chess_contest").exists():
    ROOT = ROOT.parent
    sys.path.insert(0, str(ROOT))

import chess
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import numpy as np

from chess_contest.stigmergy.fields import build_diffused
from chess_contest.stigmergy.weights import CHANNEL_NAMES, default_weights, load_weights

wpath = ROOT / "chess_contest/weights/base_weights.json"
weights = load_weights(wpath) if wpath.exists() else default_weights()
board = chess.Board()
fw, fb, aux = build_diffused(board, weights.field, weights.diffusion_steps)
print("material", aux["material"], "channels", fw.shape[0])
```

```python
fig, axes = plt.subplots(2, 5, figsize=(12, 5))
for i, ax in enumerate(axes.flat):
    field = fw[i] - fb[i]
    im = ax.imshow(field, cmap="coolwarm")
    ax.set_title(CHANNEL_NAMES[i], fontsize=8)
    ax.set_xticks([]); ax.set_yticks([])
fig.colorbar(im, ax=axes.ravel().tolist(), fraction=0.02)
fig.suptitle("White − Black diffused pheromone (startpos)")
plt.tight_layout()
out = ROOT / "chess_contest/weights/field_preview.png"
fig.savefig(out, dpi=120)
print("wrote", out)
```
