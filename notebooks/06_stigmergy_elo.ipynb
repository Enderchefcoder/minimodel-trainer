# 06 — Stigmergy Elo + contest composite

Estimate Elo on the reference ladder and combine with uniqueness scoring.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().resolve()
if not (ROOT / "chess_contest").exists():
    ROOT = ROOT.parent
    sys.path.insert(0, str(ROOT))

from chess_contest.stigmergy.elo import estimate_elo, save_report
from chess_contest.stigmergy.uniqueness import composite_contest_score, score_uniqueness
from chess_contest.stigmergy.weights import default_weights, load_weights

wpath = ROOT / "chess_contest/weights/base_weights.json"
weights = load_weights(wpath) if wpath.exists() else default_weights()
report = estimate_elo(weights, games_per_opponent=4, move_time_ms=150, max_depth=3, max_plies=60)
uniq = score_uniqueness(weights.to_dict()["uniquenessFingerprint"])
comp = composite_contest_score(report.engine_elo, bracket_winrate=0.65, uniqueness=uniq.score)
print("Elo", report.engine_elo)
print("Uniqueness", uniq.score)
print("Composite", comp)
save_report(report, ROOT / "chess_contest/weights/elo_report.json")
```
