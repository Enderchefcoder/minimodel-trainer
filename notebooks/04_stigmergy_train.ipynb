# 04 — Train Stigmergy (Diffusive Pheromone Field Engine)

Train the contest chess engine on CPU (or CUDA if available), then export JSON weights for the HTML player.

**Architecture:** multi-channel pheromone diffusion + bilinear interactions + ternary ant-trail book — not Stockfish/NNUE/AlphaZero.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().resolve()
if (ROOT / "chess_contest").exists():
    pass
elif (ROOT.parent / "chess_contest").exists():
    ROOT = ROOT.parent
    sys.path.insert(0, str(ROOT))
print("ROOT", ROOT)

import torch
from chess_contest.stigmergy.train import TrainConfig, train_base
from chess_contest.stigmergy.uniqueness import score_uniqueness

device = "cuda" if torch.cuda.is_available() else "cpu"
print("device:", device)
```

```python
cfg = TrainConfig(
    seed=42,
    imitation_steps=200,
    selfplay_rounds=24,
    games_per_round=4,
    move_time_ms=80,
    max_depth=3,
    max_plies=60,
    device=device,
    out_path=str(ROOT / "chess_contest/weights/base_weights.json"),
)
# For a quick smoke run, shrink the knobs above (or pass --quick on the CLI).
weights = train_base(cfg)
print(weights.training_meta)
print("uniqueness", score_uniqueness(weights.to_dict()["uniquenessFingerprint"]).score)
```
